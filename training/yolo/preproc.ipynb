```python
import os

DATA_PATH = 'data/selfdriving/'
LABELS_PATH = os.path.join(DATA_PATH, 'labels')
IMAGES_PATH = os.path.join(DATA_PATH, 'images')

car_class_id = 1  # From your data.yaml, 'car' corresponds to class 1

# Iterate over all label files
for label_file in os.listdir(LABELS_PATH):
    label_path = os.path.join(LABELS_PATH, label_file)

    # Read and filter label lines
    with open(label_path, 'r') as f:
        lines = f.readlines()

    car_lines = [line for line in lines if line.startswith(f"{car_class_id} ")]

    if car_lines:
        # Overwrite with only car lines
        with open(label_path, 'w') as f:
            f.writelines(car_lines)
    else:
        # No car class found; remove label and corresponding image
        os.remove(label_path)

        # Remove corresponding image file
        image_file = os.path.splitext(label_file)[0] + '.jpg'  # or .png if your images are .png
        for ext in ['.jpg', '.png', '.jpeg']:
            image_path = os.path.join(IMAGES_PATH, os.path.splitext(label_file)[0] + ext)
            if os.path.exists(image_path):
                os.remove(image_path)
                break
```

```python
import os
import yaml

# Base dataset directory
BASE_PATH = 'data/stanford/'  # Change as needed
SPLITS = ['train', 'valid', 'test']
CLASS_TO_KEEP = 1
CLASS_NAME = 'vehicle'

def clean_labels_and_images(split):
    label_dir = os.path.join(BASE_PATH, split, 'labels')
    image_dir = os.path.join(BASE_PATH, split, 'images')

    for label_file in os.listdir(label_dir):
        label_path = os.path.join(label_dir, label_file)

        with open(label_path, 'r') as f:
            lines = f.readlines()

        # Keep only class 1 ("vehicle")
        vehicle_lines = [line for line in lines if line.startswith(f"{CLASS_TO_KEEP} ")]

        if vehicle_lines:
            # Overwrite file with only vehicle entries
            with open(label_path, 'w') as f:
                f.writelines(vehicle_lines)
        else:
            # Delete label file
            os.remove(label_path)

            # Delete corresponding image
            base_name = os.path.splitext(label_file)[0]
            for ext in ['.jpg', '.jpeg', '.png']:
                image_path = os.path.join(image_dir, base_name + ext)
                if os.path.exists(image_path):
                    os.remove(image_path)
                    break

# Clean all splits
for split in SPLITS:
    clean_labels_and_images(split)

print("Cleanup and data.yaml recreation complete.")
```

```python
import os
import yaml

# Dataset base path
BASE_PATH = 'data/v2/'  # Modify if your path differs
SPLITS = ['train', 'valid', 'test']
NEW_CLASS_ID = 1
NEW_CLASS_NAME = 'car'

def unify_classes(split):
    label_dir = os.path.join(BASE_PATH, split, 'labels')

    for label_file in os.listdir(label_dir):
        label_path = os.path.join(label_dir, label_file)

        with open(label_path, 'r') as f:
            lines = f.readlines()

        # Replace all class IDs with 1
        new_lines = []
        for line in lines:
            parts = line.strip().split()
            if len(parts) >= 5:
                # Overwrite class ID with 1, keep bbox unchanged
                parts[0] = str(NEW_CLASS_ID)
                new_lines.append(' '.join(parts) + '\n')

        # Overwrite file
        if new_lines:
            with open(label_path, 'w') as f:
                f.writelines(new_lines)
        else:
            # If somehow empty, remove label and image
            os.remove(label_path)
            base_name = os.path.splitext(label_file)[0]
            for ext in ['.jpg', '.jpeg', '.png']:
                image_path = os.path.join(BASE_PATH, split, 'images', base_name + ext)
                if os.path.exists(image_path):
                    os.remove(image_path)
                    break

# Process all splits
for split in SPLITS:
    unify_classes(split)

print("All classes unified to 'car' with class ID 1. Labels and data.yaml updated.")
```
