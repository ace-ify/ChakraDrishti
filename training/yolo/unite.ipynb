```python
import os
import shutil

def consolidate_yolo_folders(base_folders, output_folder):
    """
    Consolidates images and labels from multiple YOLO dataset folders into a single output folder.

    Args:
        base_folders (list): List of paths to the base folders containing YOLO datasets.
        output_folder (str): Path to the output folder where consolidated data will be stored.
    """
    # Define the subfolder names for images and labels
    subfolders = ['images', 'labels']
    split_folders = ['train', 'valid', 'test']

    # Create the output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)

    for split in split_folders:
        for subfolder in subfolders:
            # Create the subfolders for each split
            os.makedirs(os.path.join(output_folder, split, subfolder), exist_ok=True)

    for base_folder in base_folders:
        for split in split_folders:
            for subfolder in subfolders:
                current_folder = os.path.join(base_folder, split, subfolder)
                print(current_folder)
                if os.path.exists(current_folder):
                    # Copy files from the current folder to the consolidated output folder
                    for file_name in os.listdir(current_folder):
                        source_path = os.path.join(current_folder, file_name)
                        destination_path = os.path.join(output_folder, split, subfolder, file_name)
                        shutil.copy(source_path, destination_path)
```

```python
output_folder = 'data_united'

base_folders = [
    'data/v2',
    'data/v4',
    'data/selfdriving',
    'data/stanford',
]

consolidate_yolo_folders(base_folders, output_folder)
```

```python
import os

BASE_PATH = '' 
SPLITS = ['train', 'valid']
OLD_CLASS_ID = '1'
NEW_CLASS_ID = '0'

def update_class_ids(split):
    label_dir = os.path.join(BASE_PATH, split, 'labels')

    for file_name in os.listdir(label_dir):
        file_path = os.path.join(label_dir, file_name)

        with open(file_path, 'r') as f:
            lines = f.readlines()

        new_lines = []
        for line in lines:
            parts = line.strip().split()
            if parts and parts[0] == OLD_CLASS_ID:
                parts[0] = NEW_CLASS_ID
            new_lines.append(' '.join(parts) + '\n')

        with open(file_path, 'w') as f:
            f.writelines(new_lines)

# Apply to both train and val
for split in SPLITS:
    update_class_ids(split)

print("Class ID changes complete.")
```
